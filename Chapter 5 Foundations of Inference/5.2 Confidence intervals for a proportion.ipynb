Consider, again, the Pew Research data indicating that (from a very large sample) the proportion of Americans in favor of increasing the use of solar panel farms for energy is

$$ p = 88\%,$$

while our point estimate from a sample of 2541 Americans was

$$\hat{p} = 88.9807\%.$$

### Questions
* How certain are you that the true value of $p$ is *exactly* 88%?
* How certain are you that the value of $p$ is between 86% and 90%?
* How certain are you that the value of $p$ is between 1% and 99%?



> The wider our net, the *more confident* we can be that we'll catch $p$!
>
> But, the wider our net, the *less accurate* our estimate of $p$.



## Constructing a 95% confidence interval

A range of plausible values in which we are likely to find the population parameter is called a **confidence interval**. How likely we expect the population parameter to be in a given confidence interval is quantified by the **confidence level** of the interval.

### Example
Let's construct a **95% confidence interval** for the proportion of Americans in support of increasing the use of solar panel farms for energy.

We have a point estimate,

$$\hat{p} = 88.9807\%$$

from a sample with size $n = 2541$. We previously verfied that the sampling distribution satisfies the conditions of the Central Limit Theorem, hence it follows a normal distribution with
* $\mu = p = 0.88$
* $SE = \sqrt{\dfrac{p(1-p)}{n}} \approx 0.006.$

In other words, our point estimates should approximately follow the normal distribution
$$ N(0.88, 0.006). $$

To say that we are **95% confident** that the population parameter lies in the confidence interval is to say that, if we were to take *many* samples and determined a confidence interval for each, the true parameter value would be in about 95% of the constructed confidence intervals.

> Note: a confidence interval describes a range of plausible values for the **population parameter** — it does *not* tell us about future point estimates!

## Changing the confidence level

We can increase or decrease our confidence by altering the multiple of standard errors we use to determine the width of the confidence interval. Below are some details from page 184 of the textbook.

### Example
* Construct a 99% confidence interval for the proportion of Americans in favor of increasing the use of solar panel farms and interpret the result.
* Construct a 90% confidence interval for the proportion of Americans in favor of increasing the use of solar panel farms and interpret the result.

```python
# 99% confidence interval

# Input the values
p_hat = 0.889807
z_star = 2.58
se = 0.006

# Compute the lower and upper bounds of the confidence interval
lower = round(p_hat - z_star * se, 4)
upper = round(p_hat + z_star * se, 4)

# Print the results
print(f"99% Confidence Interval: ({lower},{upper})")
```

```python
# 90% confidence interval

# Input the values
p_hat = 0.889807
z_star = 1.65
se = 0.006

# Compute the lower and upper bounds of the confidence interval
lower = round(p_hat - z_star * se, 4)
upper = round(p_hat + z_star * se, 4)

# Print the results
print(f"90% Confidence Interval: ({lower},{upper})")
```

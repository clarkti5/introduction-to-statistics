## Confidence intervals for a difference of means

Suppose an instructor decided to run two slight variations of the same exam. Prior to passing out the exams, she shuffled the exams together to ensure each student received a random version. Summary statistics for how students performed on these two exams are shown below (from page 272 of the textbook).

Screenshot 2024-10-04 at 1.30.11 PM.png

Anticipating complaints from students who took Version B, she would like to evaluate whether the difference observed in the groups is so large that it provides convincing evidence that Version B was more diffcult (on average) than Version A.

### Exercise

Verify that proceeding with statistical inference on the difference of means using a $t$-distribution is reasonable.

### Exercise

- Create 99% confidence interval for the difference of the means from Version A and Version B of the exam.
  - $(\text{point estimate}) \pm t^* \times SE $
- Interpret the confidence interval in context and draw an appropriate conclusion.

```python
# Use the function below to check your work!

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import scipy.stats as stats

def diff_means_confidence_interval(mean_1, standard_deviation_1, sample_size_1,
                                  mean_2, standard_deviation_2, sample_size_2,
                                  confidence_level):
  t_star = stats.t.ppf(1 - (1 - confidence_level) / 2,
                       df = min(sample_size_1 - 1, sample_size_2 - 1))

  standard_error = np.sqrt(
      standard_deviation_1**2/sample_size_1 + standard_deviation_2**2/sample_size_2)

  lower = (mean_1 - mean_2) - t_star * standard_error
  upper = (mean_1 - mean_2) + t_star * standard_error

  print(f"{confidence_level * 100}% confidence interval: ({lower}, {upper})")
```

```python
diff_means_confidence_interval(79.4, 14, 30, 74.1, 20, 27, 0.99)
```

## Hypothesis tests for a differece of means

### Exercise (Example 7.30)

With a significance level of $\alpha = 0.01$, use a p-value to complete the hypothesis test.

- Create appropriate null and alternative hypotheses for the situation at hand. Note that the conditions for inference were already checked in a previous exercies.
- Find the T-score and determine the corresponding p-value.
- Using this p-value and the stated significance level, draw an appropriate conclusion.

```python
# Use the function below to check your work!

def diff_mean_t_test(mean_1, standard_deviation_1, sample_size_1,
                     mean_2, standard_deviation_2, sample_size_2,
                     null_value):
  standard_error = np.sqrt(
      standard_deviation_1**2/sample_size_1 +
      standard_deviation_2**2/sample_size_2
      )

  t_score = ((mean_1 - mean_2) - null_value)/standard_error

  p_value = (1 - stats.t.cdf(abs(t_score), df = min(sample_size_1, sample_size_2) - 1)) * 2

  print(f"T-score: {t_score}")
  print(f"p-value: {p_value}")
```

```python
diff_mean_t_test(79.4, 14, 30, 74.1, 20, 27, 0)
```

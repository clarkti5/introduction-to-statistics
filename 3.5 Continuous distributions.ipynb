Let's consider a continuous (as opposed to discrete) random variable. We'll import some summary data on the heights of men in the U.S.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd

fcid = pd.read_csv('https://raw.githubusercontent.com/clarkti5/math-209-data-sets/main/fcid.csv')
```

```python
fcid.head()
```

We can sum the `num_of_adults` variable to see how many people were sampled.

```python
fcid['num_of_adults'].sum()
```

We'll now add a column to the `fcid` data frame tha represents the proportion of adults with a given height.

```python
proportion_of_adults = fcid['num_of_adults'] / 3000000 # This creates the new column data.

fcid.insert(2, "proportion_of_adults", proportion_of_adults) # This inserts the new data as the third column.

fcid.iloc[10:20] # This will display rows 10 through 19 of the data.
```

Below are some increasingly granular histograms from this data on page 125 of the textbook.

Screenshot 2024-09-03 at 12.49.57 PM.png

Notice, as the number of bins increases (i.e., as the bin width decreases), the rectangles in the histogram almost appear to blend together into a curve. We can visualize this curve with the plot below.

```python
from plotnine import *

(
    ggplot(fcid) +
    aes(x = 'height', y = 'proportion_of_adults') +
    geom_point() +
    geom_line()
)
```

The smooth curve represents a **probability density function** (sometimes just called a **density function**).

* The height of the curve represents the "probability per unit length" of a particular input.
* The probability of observing a range of input values is represented by the **area under the probability density function**.
* The total area under the curve is 1 (i.e., 100% probability).

### Question

What is the probability that a randomly selected person's height is between 180 cm and 185 cm?

```python
heights_180_to_185 = fcid[(fcid['height'] > 180) & (fcid['height'] < 185)]

heights_180_to_185
```

```python
heights_180_to_185['num_of_adults'].sum()
```

So, the probability that a randomly selected person has a height between 180 cm and 185 cm is

```python
351546/3000000
```

Note that the total proportion of adults sampled with heights between 180 cm and 185 cm is

```python
heights_180_to_185['proportion_of_adults'].sum()
```

This could be represented on the graph of the probability density function as the area of two (very thin) rectangles approximating the true area under the curve.

In the US, elections for the House of Representatives occur every 2 years, occassionally coinciding with the Presidential election which is held every 4 years. House elections that occur during the middle of a Presidential term are called midterm elections.

A common political theory is that the economy is many voters' priority, so if the unemployment rate is too high, the sitting President's political party will do worse during the midterm elections.

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd

house = pd.read_csv('https://raw.githubusercontent.com/clarkti5/math-209-data-sets/refs/heads/main/midterms_house.csv')
house.head()
```

```python
from plotnine import *

(
  ggplot(house) +
  aes(x = 'unemp', y = 'house_change', shape = 'party') +
  geom_point() +
  labs(x = 'Unemployment rate', y = 'Percent change in seats') +
  xlim(3, 22) + ylim(-30, 10)
)
```

```python
(
  ggplot(house) +
  aes(x = 'unemp', y = 'house_change') +
  geom_point() +
  geom_smooth(method = 'lm', se = False) +
  labs(x = 'Unemployment rate', y = 'Percent change in seats') +
  xlim(3, 22) + ylim(-30, 10)
)
```

- Use the model to predict the percent change in House seats for an unemployment rate of 6.1%.
- In 1995, Bill Clinton's party lost about 21% of their House seats with an unemployment rate of 6.1%. Is the residual of the model positive or negative here?
- What percent of the House seats change is explained by the unemployment rate, according to the model?
- What is the correlation coefficient of the model?

```python
# Import the `statsmodels` library
import statsmodels.api as sm

# Set the formula string (which variables to compare) and the dataframe to use
formula_string = "house_change ~ unemp"
dataframe = house

# Fit the linear model
model = sm.formula.ols(formula = formula_string, data = dataframe)
model_fitted = model.fit()

# Print the results
print(model_fitted.summary())
```

#### Solution

- From the regression output, the model is

$$\hat{\texttt{house_change}} = -7.3644 - 0.8897 \cdot \texttt{unemp}.$$

- **Slope.** The model predicts that an increase of $1\%$ in unemployment will result in a decrease of $0.8897\%$ in House seats for the sitting president's political party on average.

- If `unemp` is $6.1\%$, we have
$$ \hat{\texttt{house_change}} = -7.3644 - 0.8897 \cdot 6.1 \approx -12.7916\%.$$

- Since the predicted value is greater than the observed value, the residual at $6.1\%$ unemployment is negative.

- Since $R^2 = 0.040$, about $4\%$ of the variation in House seats change is explained by unemployment.

- We have $R^2 = 0.040$, so we must have $R = \pm \sqrt{0.040} = \pm 0.2$. Since the model has a negative slope, $R = -0.20$.

### Question
How can we tell if unemployment is *actually* a good predictor of the change in House seats?

## Hypothesis testing for the slope of a linear model

Consider the following hypotheses.

- $H_0$: The slope of the true linear model is 0. That is to say, unemployment rate is *not* predictive of the change in House seats.
- $H_A$: The slope of the true linear model is not 0. That is, unemployment rate is predictive of the change in House seats.

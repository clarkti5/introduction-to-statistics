Sometimes, the relationship between two numerical variables appears roughly *linear*, meaning that data points appear to be scattered around a line. Linear functions are particularly well-behaved, so fitting a linear model to your data can be very advantageous.

```python
import warnings
warnings.filterwarnings('ignore')

import pandas as pd

possums = pd.read_csv('https://raw.githubusercontent.com/clarkti5/math-209-data-sets/refs/heads/main/possum.csv')
possums.head()
```

## Fitting a linear model

Below is a scatter plot visualizing the relationship between total length (`total_l`, in cm) and head length (`head_l`, in mm).

```python
from plotnine import *

plot = (
  ggplot(possums, aes(x='total_l', y='head_l')) +
  geom_point()
)

plot
```

## Residuals

The difference between the observed value and the predicted value is called a **residual**. In other words,

$$\text{Residual} = \text{Observed} - \text{Predicted}.$$

## Correlation

The **correlation** (or **correlation coefficient**) is a number between -1 and 1 that is used to quantify the strength of a linear relationship. Technically, it is computed as

$$ R = \frac{1}{n-1}\sum_{i = 1}^n \frac{x_i - \bar{x}}{s_x} \cdot \frac{y_i - \bar{y}}{s_y}. $$

We'll typically used technology to compute $R$ for us. Here are some examples from page 311 of the textbook.

$R$ is meant to help us quantify the strength of *linear* relationships. Nonlinear trends, even when very strong, can produce $R$ values that suggest relatively weak relationship. Always visualize your data!

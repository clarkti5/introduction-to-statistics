Recall our conditions for proceeding with linear regression.

- **Linearity.** The data should follow a linear trend.

- **Nearly normal residuals.** The residuals should be nearly normal without extreme outliers.

- **Constant variability.** The variability of points around the regression line should be roughly constant.

- **Independent observations.** As usual, we need observations to be independent (e.g., from a random sample).

The *nearly normal residuals* condition involves identifying outliers along with their impact on the resulting model.

Points that are a large horizontal distance away from the point cloud are called **high leverage** points --- they tend to "pull" harder on the line than other points. If a point actually ends up altering the model significantly (as the highe leverage points in (3), (4), and (5) do), they are called **influential points**.

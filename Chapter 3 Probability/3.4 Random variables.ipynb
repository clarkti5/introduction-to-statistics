### Expected value of a random variable
For a random variable $X$ with outcomes $x_i$, the **expected value** of $X$ is the weighted average of its outcomes with weights given by the probability of each outcome.

## Linear combinations of random variables

Sometimes, we can break down a random variable into a **linear combination** of other (hopefully simpler) random variables. Here, "linear combination" refers to a combination resulting from the use of addition and multiplication by a real number. In symbols, a linear combination of random variables $X$ and $Y$ will look like

$$ aX + bY$$

where $a$ and $b$ are real numbers.

Let $X$ continue to represent the amount of money spent by a single student on statistics textbooks.

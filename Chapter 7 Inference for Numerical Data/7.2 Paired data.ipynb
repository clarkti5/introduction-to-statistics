Two data sets are called **paired** if each observation in one set has a special correspondence with *exactly one* obesrvation in the other set.

An example is contained in the `textbooks` data set. In this data set is a list of textbooks with their prices at the UCLA Bookstore as well as on Amazon. Since the textbooks are the *same*, the UCLA Bookstore price data is paired with the Amazon price data.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
import numpy as np
import scipy.stats as stats

textbooks = pd.read_csv('https://raw.githubusercontent.com/clarkti5/math-209-data-sets/refs/heads/main/textbooks.csv')
textbooks = textbooks.drop(columns = 'more')
textbooks.head()
```

### Question

Can we see if the UCLA Bookstore charges a different price than Amazon for textbooks, on average?

### Exercise (Example 7.18)

Complete the hypothesis test.

- Compute the standard error and T-score for the mean of the price differences.
- Determine a p-value using a $t$-distribution with the appropriate degrees of freedom.
- Use a significance level of $\alpha = 0.05$ to draw an appropriate conclusion.

```python
# Check your result with this function
# Paired mean t-test

def mean_t_test(mean, standard_deviation, sample_size, null_value):
  standard_error  = standard_deviation / np.sqrt(sample_size)

  t_score = (mean - null_value)/standard_error

  p_value = (1 - stats.t.cdf(abs(t_score), df = sample_size - 1)) * 2

  print(f"T-score: {t_score}")
  print(f"p-value: {p_value}")
```

```python
mean_t_test(12.761644, 14.255301, 73, 0)
```

A simple random sample of 826 payday loan borrowers was surveyed to better understand their interests around regulation and costs. 70% of the responses supported new regulations on payday lenders.

- **Confidence interval.** Use $\hat{p} = 0.70$ to construct a 95% confidence interval for $p$, the proportion of payday borrowers who support increased regulation for payday lenders.
  - Interpret the confidence interval in context.
  - In particular, does this survey provide evidence that the majority of borrowers support or oppose these new regluations?
- **Hypothesis test.** Set up and test hypotheses to evaluate whether borrowers have a majority support or majority opposition for new regulations on payday lenders.
  - Interpret the conclusion in context.
  - Compare your conclusion to the results of the confidence interval previously.

```python
# Use the functions below to check your work.
import warnings
warnings.filterwarnings('ignore')

import numpy as np
import scipy.stats as stats
import pandas as pd
import statsmodels.api as sm

# Single proportion confidence interval
def proportion_confidence_interval(proportion, sample_size, confidence_level):
  z_star = stats.norm.ppf(1 - (1 - confidence_level) / 2)

  standard_error = np.sqrt(proportion * (1 - proportion)/sample_size)

  lower = proportion - z_star * standard_error
  upper = proportion + z_star * standard_error
  print(f"{confidence_level * 100}% confidence interval: ({lower}, {upper})")

# Single proportion z-test
def proportion_z_test(proportion, sample_size, null_value):
  standard_error = np.sqrt(null_value * (1 - null_value) / sample_size)

  z_score = (proportion - null_value) / standard_error

  p_value = (1 - stats.norm.cdf(abs(z_score))) * 2

  print(f"Z-score: {z_score}")
  print(f"p-value: {p_value}")
```

```python
proportion_confidence_interval(0.7, 826, 0.95)
```

```python
proportion_z_test(0.7, 826, 0.5)
```

## Choosing a sample size

The **margin of error** for a confidence interval is

$$z^* \times SE = z^* \times \sqrt{\dfrac{p(1-p)}{n}}. $$

## Getting started with data

The Behavioral Risk Factor Surveillance System (BRFSS) is an annual telephone survey of 350,000 people in the United States. As its name implies, the BRFSS is designed to identify risk factors in the adult population and report emerging health trends. For example, respondents are asked about their diet and weekly physical activity, their HIV/AIDS status, possible tobacco use, and even their level of healthcare coverage. The BRFSS's [website](http://www.cdc.gov/brfss) contains a complete description of the survey, including the research questions that motivate the study and many interesting results derived from the data.

We will focus on a random sample of 20,000 people from the BRFSS survey conducted in 2000. While there are over 200 variables in this data set, we will work with a small subset.

We begin by importing the dataset of 20,000 observations from the Cloud.

````python
# These `import` lines bring in extra functionality to Python that we'll need to do statistics.

import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import io
import requests

# These lines pull the data from the cloud and store it as a dataframe named `cdc`.

df_url = 'https://raw.githubusercontent.com/akmand/datasets/master/openintro/brfss_2000.csv'
url_content = requests.get(df_url, verify=False).content
cdc = pd.read_csv(io.StringIO(url_content.decode('utf-8')))
cdc.shape
````

````python
# The .sample() function takes a random sample from the `cdc` data.

cdc.sample(10, random_state=999)
````

The data set `cdc` that shows up is a data matrix, with each row representing a case and each column representing a variable. This kind of data format are called **dataframe**, which is a term that will be used throughout the labs.

We can see the column names are `genhlth`, `exerany`, `hlthplan`, `smoke100`, `height`, `weight`, `wtdesire`, `age`, and `gender`. Each one of these variables corresponds to a question that was asked in the survey.

- For example, for `genhlth`, respondents were asked to evaluate their general health, responding either excellent, very good, good, fair or poor.

- The `exerany` variable indicates whether the respondent exercised in the past month (`1`) or did not (`0`).

- The `smoke100` variable indicates whether the respondent had smoked at least 100 cigarettes in their lifetime.

- The other variables record the respondent's `height` in inches, `weight` in pounds as well as their desired weight, `wtdesire`, `age` in years, and `gender`.

## Summaries for numerical variables

The BRFSS questionnaire is a massive trove of information. A good first step in any analysis is to distill all of that information into a few summary statistics and graphics. As a simple example, the function `.describe()` returns a numerical summary: count, mean, standard deviation, minimum, first quartile, median, third quartile, and maximum. The code below finds a numerical summary for `weight`.

````python
# This code finds a numerical summary for `weight`.

cdc['weight'].describe()
````

If you wanted to compute the interquartile range for the respondents' weight, you could look at the output from the summary command above and tell Python to do the arithmetic for you.

Python also has built-in functions to compute summary statistics one by one.

````python
# This code prints the numerical summary items one line at a time.

print('count:', cdc['weight'].count())
print('mean: ', cdc['weight'].mean())
print('standard deviation:  ', cdc['weight'].std())
print('variance:  ', cdc['weight'].var())
print('minimum:  ', cdc['weight'].min())
print('25th percentile:  ', cdc['weight'].quantile(0.25))
print('50th percentile:  ', cdc['weight'].median())
print('75th percentile:  ', cdc['weight'].quantile(0.75))
print('maximum:  ', cdc['weight'].max())
````

## Visualizing categorical data

Now let's import the `plotnine` library to create plots. This library is based on the grammar of graphics paradigm used by `ggplot` in the `R` programming language.

````python
from plotnine import *
````

You can also change the default style of plots. Let's go for our favourite style, `tufte`, which is a minimalist theme based on the design principles of Edward Tufte.

````python
theme_set(theme_tufte())
````

Now we can make a bar plot of the entries in the table. In the first line, we use

`cdc.replace([0,1], ["0","1"])`

as the dataframe. This replaces instances of the *numbers* $0$ and $1$ with text *strings* "0" and "1". Otherwise, Python confuses the categorical variable `smoke100` for a numerical variable and the plot is not displayed correctly. In general, Python thinks about numbers and strings differently, so we occasionally need to be careful about which one we're using.

````python
(
  ggplot(cdc.replace([0,1],["0","1"])) + # We tell `ggplot()` what dataframe to use.
  aes(x = 'smoke100') +                  # Describe the variable(s).
  geom_bar() +                           # Identify the type of plot.
  labs(title = 'Bar plot of smoke100')   # Set the plot title.
)
````

Notice how the code builds the plot one line at a time, using a `+` to introduce new features one at a time.

The `.crosstab()` command can be used to tabulate and compare counts across more than one categorical variable. For example, to examine which participants have smoked across each gender, we could use the following code.

````python
pd.crosstab(cdc['gender'], cdc['smoke100'])
````

The `.crosstab()` function also allows us to include "margins" in our table for totals.

````python
pd.crosstab(cdc['gender'], cdc['smoke100'], margins = True, margins_name = 'Total')
````

If we'd rather see proportions instead of counts, we can include the `normalize = True` option, similar to the `.value_counts()` function.

````python
pd.crosstab(cdc['gender'], cdc['smoke100'], margins = True, margins_name = 'Total', normalize = True)
````

In these tables, we see column labels of 0 and 1. Recall that 1 indicates a respondent has smoked at least 100 cigarettes. The rows refer to gender.

We can visualize this table with a mosaic plot. To create a mosaic plot, we first define a function called `mosaic()` the will make a plot based on three inputs:

- `data`: the data to be used,

- `var1`: the horizontal variable,

- `var2`: the vertical variable.

Functions are a powerful tool in Python that allows us to easily reuse complicatd code structures. Just run the code below to define the `mosaic()` function.

````python
def mosaic(data, var1, var2):
  # Group by 'var1' and 'var2', then summarize counts
  df = data.groupby([var1, var2]).size().reset_index(name='count')

  # Calculate the total count per 'var1'
  df['var1_count'] = [df[df[var1] == (df[var1][i])]['count'].sum() for i in range(len(df))]

  # Calculate the proportion of each count
  df['prop'] = df['count'] / df['var1_count']

  plot = (
  ggplot(df, aes(x = var1, y = 'prop', width = 'var1_count', fill = var2)) +
  geom_bar(stat = "identity", position = "fill", colour = "black") +
  facet_grid(cols = var1, scales = "free_x", space = "free_x") +
  theme(axis_text = element_blank(), axis_title = element_blank(), axis_ticks = element_blank())
  )

  return plot
````

Now, we can run the `mosaic()` to make a mosaic plot of `smoke100` across `gender`.

````python
mosaic(cdc.replace([0,1], ["0","1"]), 'gender', 'smoke100')
````

### A little more on subsetting

It's often useful to extract all individuals (cases) in a data set that have specific characteristics. We accomplish this through conditioning commands. First, consider expressions like the following.

````python
cdc['gender'] == 'm'
````

````python
cdc['age'] > 30
````

These commands produce a series of `TRUE` and `FALSE` values. There is one value for each respondent, where `TRUE` indicates that the person was male (via the first command) or older than 30 (second command).

Suppose we want to extract just the data for the men in the sample, or just for those over 30. For example, the following command will create a new dataframe called `mdata` that contains only the men from the `cdc` data set.

````python
mdata = cdc[cdc['gender'] == 'm']
````

The syntax here may look a little confusing at first, so let's break it down.

- `cdc` is the name of our dataframe.
- Typing `cdc[???]` will tell Python to look into the `cdc` dataframe and find whatever `???` tells it to look for.
- We want it to look for only the entries of male correspondents, i.e., where the `gender` variable is `'m'`.
- Python uses `==` to say "is equal to." Since `cdc['gender']` refers to the `gender` column of `cdc`, the code `cdc['gender'] == 'm'` will tell Python to find where the `gender` variable is `'m'`.
- Putting it altogether, `cdc[cdc['gender'] == 'm']` tells Python to look into the `cdc` dataframe and return with only the entries where the `gender` column has value `'m'`.

You can take a peek at the first several rows as usual with the `.head()` command.

````python
mdata.head()
````

We can carve up the data based on values of one or more variables. We can use several of these conditions together with `&` and `|`. The `&` is read "and" so that the following code will give you the data for men over the age of 30.

````python
m_and_over30 = cdc[(cdc['gender'] == 'm') & (cdc['age'] > 30)]
m_and_over30.head()
````

The `|` character is read "or" so that the following will will take people who are men or over the age of 30 (why that's an interesting group is hard to say, but right now the mechanics of this are the important thing).

````python
m_or_over30 = cdc[(cdc['gender'] == 'm') | (cdc['age'] > 30)]
m_or_over30.head()
````

## Visualizing numerical data

With our subsetting tools in hand, we'll now return to the task of the day: making basic summaries of the BRFSS questionnaire. We've already looked at categorical data such as `smoke100` and `gender` so now let's turn our attention to quantitative data. Two common ways to visualize quantitative data are with box plots and histograms. We can construct a box plot for a single variable with the following command.

````python
(
  ggplot(cdc) +
  aes(x = [''], y = 'height') +
  geom_boxplot(width = 0.25) + # Using `geom_boxplot()` tells Python to make a boxplot.
  labs(title = 'Boxplot of height')
)
````

We can compare the locations of the components of the box by examining the summary statistics.

````python
cdc['height'].describe()
````

Confirm that the median and upper and lower quartiles reported in the numerical summary match those in the graph.

The purpose of a boxplot is to provide a sketch of a variable for the purpose of comparing across several categories. For example, we can compare the heights of men and women with the code below.

````python
(
  ggplot(cdc) +
  aes(x = 'gender', y = 'height') +
  geom_boxplot(width = 0.25) +
  labs(title = 'Boxplot of height by gender')
)
````

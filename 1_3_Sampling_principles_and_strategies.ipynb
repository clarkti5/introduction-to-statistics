## Populations and samples

### Question

> What proportion of Michigan highschool graduates are ready for college-level math?

Every research question refers to a **target population**. A **sample** represents a subset of the cases and is often a small fraction of the population.

Ideally, a sample should be

- **Representative** of the population, so that results can be **generalized** to the population.
- Large enough to avoid **anecdotal evidence**.

We'll use some data on college readiness for Michigan highschool students collected from 2017-2018.

```python
import warnings
warnings.filterwarnings("ignore")

import pandas as pd
from plotnine import *

college_ready = pd.read_csv('https://raw.githubusercontent.com/clarkti5/math-209-data-sets/refs/heads/main/CollegeReadiness_2017_2018_byZip_20181107.csv')
```

```python
college_ready.head()
```

```python
college_ready[['MathNumAssessed', 'MathNumReady']].describe()
```

The data is broken down by ZIP code in the `GEOID10` column. With the wonders of technology, we can compute the proportion of students who are college ready in math directly.

```python
df = college_ready.dropna()

sum(df['MathNumReady'])/sum(df['MathNumAssessed'])
```

Next, we'll compare this number to the average proportion of college ready students from a sample.

We can use Python to help us take a **random sample**, which is essentially equivalent to drawing names from a hat.

```python
college_ready_sample = df.sample(n=30)
college_ready_sample.head()
```

The next code will produce the proportion of students who are college ready in math from our sample.

```python
sum(college_ready_sample['MathNumReady'])/sum(college_ready_sample['MathNumAssessed'])
```

Generally speaking, we expect a sample to vary slightly from the target population. When computing the proportion of math readiness from the sample, we are computing a **sample statistic** (sometimes called a **point estimate**). Sample statistics are intended to approximate a **population parameter**, which in this case would be the proportion of *all* Michigan highschool graduates who are college ready in mathematics.

## Four sampling methods

We will explore four primary sampling methods.

- **Simple random sampling** chooses members of the sample from the population randomly.

- **Stratified sampling** first divides the population into groups called **strata** and then takes a simple random sample from each strata.

- **Clustering** first divides the population into groups called **clusters**, and then sample *all* observations from a few of the clusters.

- **Multistage clustering** first divides the population into groups called **clusters**, and then uses a different sampling technique on a few of the clusters.

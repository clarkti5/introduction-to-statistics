Studies where the researchers assign treatments to cases are called **experiments**. When this assignment includes randomization, e.g. using a coin flip to decide which treatment a patient receives,
it is called a **randomized experiment**. Randomized experiments are fundamentally important when trying to show a **causal connection** between two variables.

## Principles of experiment design

There are 4 primary principles to the design of experiments.

- **Control.** Researchers assign treatment(s) to cases/patients, doing their best to control any differences between the groups.
  - Commonly, cases are placed into a **treatment group** and a **control group**. The control group may recieve a **placebo** treatement.

- **Randomization.** Researchers randomize placement into treatment groups to account for variables that cannot be controlled.

- **Replication.** The more cases observed (either in one or multiple studies), the more accurately we can estimate the effect of the explanatory variable on the response variable.

- **Blocking.** This refers to grouping cases into blocks based on variables (other than the explanatory variable) tha may influence the response variable. By first splitting cases into blocks, it ensures an equal representation of the blocks in the treatment groups.
  - Common blocks are for low-risk and high-risk patients. Blocking is not a typical feature of experiments for our purposes.

## Reducing bias in human experiments

Experiments involving human participants must deal with the difficult to quantify *emotional* effect of giving or receiving a treatment.

- When researchers keep the patients uninformed about their treatment, the study is said to be **blind**. Placebos (fake treatments) are often used to maintain blindness. Often times, a placebo results in a slight but real improvement in patients. This effect has been
dubbed the **placebo effect**.

- Most modern studies are **double-blind**, meaning the patients *as well as* the doctors, researchers, or *anyone else* interacting with the patients are unaware of who is or is not receiving the treatment.

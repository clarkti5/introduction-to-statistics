Researchers recruited 219 participants in a study where they would like to sell a used iPod that was known to have frozen twice in the past. The participants were incentivized to get as much money as they could for the iPod since they would
receive a 5% cut of the sale on top of $10 for participating. The researchers wanted to understand what types of questions would elicit the seller to disclose the freezing issue.

The buyers were collaborating with the researchers and were scripted to ask

> "Okay, I guess I'm supposed to go first. So, you've had the iPod for 2 years...

followed by one of the following questions.

- General/Neutral: What can you tell me about it?
- Positive Assumption: It doesn't have any problems, does it?
- Negative Assumption: What problems does it have?

We'd like to know if the question asked influenced the sellers to disclose the iPod's freezing issues. The results are summarized in the two-way table below from page 240 of the textbook.

## Chi-square test for two-way tables

With two-way tables, we are generally interested in questions about *independence* between the variables. The table seems to suggest that the sellers disclosure is *not* independent of the question asked.

### Exercise

- Construct null and alternative hypothesese for this situation.
- Assuming the null hypothesis, what proportion of sellers will disclose the iPod's freezing problem?
- Under this assumption, calculate the *expected* counts for each cell in the two-way table above.
- How do the observed counts compare to the expected counts?

#### Solution

- $H_0$: Disclosure is *independent* of the question asked.
- $H_A$: Disclosure is *not* independent of the question asked.

Assuming $H_0$, the overall proportion of disclosure should be

$$ \frac{61}{219} \approx 0.2785. $$

Since there were 73 people in the **General** group, the expected number of disclosures in the **General** group is

$$ 0.2785 \cdot 73 \approx 20.33.$$

In fact, this is the same across all groups, since each group had 73 cases. In general, the expected count for a given cell will look like

$$\boxed{\frac{\text{row total}}{\text{table total}}} \cdot \boxed{\text{column total}}.$$

Assuming $H_0$, the overall proportion of non-disclosure (i.e. hiding the problem) should be

$$ \frac{158}{219} \approx 0.7215. $$

Since there were 73 cases in each group, the expected number of non-disclosures is

$$ 0.7215 \cdot 73 \approx 52.67.$$

### Exercise

Compute the $\chi^2$ test statistic for this two-way table by finding the sum of

$$ \dfrac{(\text{observed count} - \text{expected count})^2}{\text{expected count}} $$

for each cell in the two-way table.

```python
# Calculate the chi-square value

(2 - 20.33)**2/20.33 + (23 - 20.33)**2/20.33 + (36 - 20.33)**2/20.33 + (71 - 52.67)**2/52.67 + (50 - 52.67)**2/52.67 + (37 - 52.67)**2/52.67
```

### Exercise

Similar to Section 6.3, this $\chi^2$ value will follow a chi-square distribution. However, since there are two variables instead of one, the degrees of freedom for the distribution must be computed as

$$(\text{number of rows - 1}) \times (\text{number of columns - 1}). $$

Find the degrees of freedom for our two-way table above.

```python
# Find the degrees of freedom

(2 - 1) * (3 - 1)
```

### Exercise

We found $\chi^2 = 40.13$ with 2 degrees of freedom. [Determine a p-value](https://geogebra.org/classic#probability) and (using a significance level of $\alpha = 0.05$) draw an appropriate conclusion regarding whether the question asked affects the sellers liklihood of disclosing the freezing problem.

```python
# Run this code to check the results of the exercises above.

import scipy.stats as stats
import numpy as np

observations = np.array([[2, 23, 36], [71, 50, 37]])

chi_square_test = stats.chi2_contingency(observations)

print(f"chi-square value = {chi_square_test.statistic}")
print(f"p-value = {chi_square_test.pvalue}")
```
